# Ejercicio 6 - Limpiar la serie horaria
inicia declarando las 12 lecuras en lista y se recorre con un for luego se calcula el promedio, las lecturas descartadas y la cantidad de lecturas validas.

```python
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]
lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

for lectura_actual in lecturas_pm25:
    if lectura_actual == -999.0:
        lecturas_descartadas += 1
        continue
    lecturas_validas.append(lectura_actual)
    suma_pm25 += lectura_actual

promedio_pm25 = suma_pm25 / len(lecturas_validas)
print("Lecturas validas:",len(lecturas_validas))
print("Lecturas descartadas:",lecturas_descartadas)
print("Promedio:",promedio_pm25)
```

# Ejercicio 7 — Estadística descriptiva de la serie limpia
Se calculea el resto de la descripción de la serie. Necesita dos recorridos: el primero para los extremos, el segundo para la dispersión (que depende del promedio, y por eso no puede calcularse en la misma pasada).
Entonces, se inicializa variables y se recorre con for e if, se calcula el rango restando el mínimo al máximo y se acumula la diferencia entre cada lectura y la cantidad de lecturas validas, como resultado se obtiene el minimo, maximo, rango y la desviación

```python
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

for valor in lecturas_validas:
    if valor > maximo_pm25:
        maximo_pm25 = valor
    if valor < minimo_pm25:
        minimo_pm25 = valor

rango = maximo_pm25 - minimo_pm25

suma_desviaciones = 0.0
for valor in lecturas_validas:
    suma_desviaciones += (valor - promedio_pm25) ** 2

desviacion_estandar = (suma_desviaciones / len(lecturas_validas)) ** 0.5

print("Maximo:",maximo_pm25)
print("Minimo:",minimo_pm25)
print("Rango:",rango)
print("Desviacion:",desviacion_estandar)
```

# Ejercicio 8 — Decaimiento tras cerrar la fuente de emisión
El codigo es saber cuántas horas tardará en volver a categoría "Buena", se declara y se usa while para repetir y actualizar para no ghenerar un bucle infinito y luego se muestra cuantas horas pasaron y con qué concentración final se cerró el episodio.

```python
concentracion = 58.3
factor_decaimiento = 0.85
umbral_buena = 12.0

horas_transcurridas = 0

while concentracion > umbral_buena:
    concentracion = concentracion * factor_decaimiento
    horas_transcurridas += 1

print("Horas transcurridas;",horas_transcurridas)
print("Concentracion:",concentracion)
```

# Ejercicio 9 — Registros de varias estaciones
Se recorre la lista de diccionarios y cada una tendra la tupla de coordenadas, se calcula el promedio y se obtiene la humedad con get (para no generar KeyError) y se clasifica el promedio con if, elif, else.

```python
registros_estaciones = [
    {"codigo": "EST-01", "sector": "Norte",  "coordenadas": (6.28, -75.57), "lecturas": [8.4, 12.7, 15.2]},
    {"codigo": "EST-02", "sector": "Centro", "coordenadas": (6.24, -75.58), "lecturas": [41.8, 58.3, 36.1]},
    {"codigo": "EST-03", "sector": "Sur",    "coordenadas": (6.19, -75.59), "lecturas": [9.6, 13.9, 7.5], "humedad_relativa": 68.0},
]

for estacion in registros_estaciones:
    latitud_estacion, longitud_estacion = estacion["coordenadas"]

    suma_lecturas_estacion = 0.0
    for valor in estacion["lecturas"]:
        suma_lecturas_estacion += valor
    promedio_estacion = suma_lecturas_estacion / len(estacion["lecturas"])

    humedad_estacion = estacion.get("humedad_relativa", "no disponible")

    if promedio_estacion <= 12.0:
        categoria_estacion = "Buena"
    elif promedio_estacion <= 35.4:
        categoria_estacion = "Moderada"
    elif promedio_estacion <= 55.4:
        categoria_estacion = "Dañina para grupos sensibles"
    else:
        categoria_estacion = "Dañina para la salud"

    print(
        "Código:", estacion["codigo"],
        "| Sector:", estacion["sector"],
        "| Latitud:", latitud_estacion,
        "| Longitud:", longitud_estacion,
        "| Promedio:", round(promedio_estacion, 2),
        "| Categoría:", categoria_estacion,
        "| Humedad:", humedad_estacion,
    )
```
